```python
import pandas as pd
```

```python
pd.set_option("display.max_columns", None)  # show all cols
pd.set_option("display.max_colwidth", None)  # show full width of showing cols
pd.set_option(
    "display.expand_frame_repr", False
)  # print cols side by side as it's supposed to be
```

```python
df = pd.read_parquet("ofgl.parquet", engine ="auto")
df.columns
```

```python
df["agregat"].unique()
```

```python
impots_locaux = df[df["agregat"] == "Impôts locaux"]
impots_locaux.head()
```

```python
verif = impots_locaux[impots_locaux["com_code"] == "01263"]
verif = verif.sort_values("exer", ascending = False)
verif.head(1)
```

```python
impots_locaux["exer"].value_counts()
```

```python
impots_locaux["annee"] = impots_locaux["exer"].astype(str).str[:4].astype(int)
impots_locaux["annee"].value_counts()
```

```python
impots_locaux = impots_locaux[["annee", "com_code", "com_name", "agregat", "montant", "ptot", "euros_par_habitant"]]
impots_locaux.head()
```

```python
impots_locaux = impots_locaux.rename(columns = {
    "com_code" : "code_geo"
})
impots_locaux.head(2)
```

```python
#export en CSV
impots_locaux.to_csv("ofgl_impots_locaux.csv", index=False)
```
