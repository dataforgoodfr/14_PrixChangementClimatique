```python
# Importation des bibliothèques nécessaires
import duckdb
import pandas as pd

# Connexion à la base dans le dossier exploration
con = duckdb.connect("dev.duckdb")
```

```python
# Vérification de la connexion en listant les tables existantes
con.execute("SHOW TABLES").fetchdf()
```

```python
# Création du schéma dédié pour REI
con.execute("CREATE SCHEMA IF NOT EXISTS rei;")
```

```python
# Chargement du fichier nettoyé dans la base
clean_csv_path = "C:\\Users\\thier\\Desktop\\rei\\csv_clean\\rei_clean.csv"

con.execute(f"""
    CREATE OR REPLACE TABLE rei.rei_clean AS
    SELECT *
    FROM read_csv_auto('{clean_csv_path}')
""")
```

```python
# Vérification de la table ingérée
con.execute("SELECT * FROM rei.rei_clean LIMIT 10").fetchdf()
```

```python
# Compte le nombre de lignes
con.execute("SELECT COUNT(*) FROM rei.rei_clean").fetchall()
```

```python
# Affiche le type de colonnes
con.execute("DESCRIBE rei.rei_clean").fetchdf()
```

```python
# Vérification du taux de valeurs manquantes
df = con.execute("SELECT * FROM rei.rei_clean").fetchdf()
missing_rate = df.isna().mean() * 100
missing_rate
```
