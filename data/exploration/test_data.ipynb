```python
import sqlite3
import pandas as pd

# connexion SQLite en mémoire
conn = sqlite3.connect(":memory:")

# envoyer ton dataframe dans SQLite
budget_per_compte.to_sql("budget_insee", conn, index=False, if_exists="replace")
sireninsee.to_sql("sireninsee", conn, index=False, if_exists="replace")
# requête SQL
query = """
WITH transformed AS (SELECT
    -- Transformation du code département
    CASE
        WHEN LENGTH(printf('%02d', CAST(NDEPT AS INTEGER))) = 3
             AND SUBSTR(printf('%02d', CAST(NDEPT AS INTEGER)), 1, 2) = '10'
        THEN '97'
        ELSE printf('%02d', CAST(NDEPT AS INTEGER))
    END AS code_departement,

    -- Transformation du code INSEE
    CASE
        WHEN SUBSTR(printf('%02d', CAST(NDEPT AS INTEGER)), 1, 2) = '97'
        THEN SUBSTR(printf('%02d', CAST(NDEPT AS INTEGER)), -1, 1) || printf('%02d', INSEE)
        ELSE printf('%03d', INSEE)
    END AS code_insee,
    IDENT AS siret,
    SUBSTR(IDENT, 1, 9) AS siren,
        EXER AS annee,
    CREGI AS code_region,
    type_compte AS type_compte,
    SD AS solde_debiteur,
    SC AS solde_crediteur
FROM budget_insee)
SELECT
    t.*,
    t.code_departement || t.code_insee AS code_geo,
    s.insee AS insee_siren,
    s.nom_com
FROM transformed t
LEFT JOIN sireninsee s
    ON t.siren = s.siren;
"""

result = pd.read_sql(query, conn)
result.head()
```
